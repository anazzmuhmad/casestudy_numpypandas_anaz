```python
import pandas as pd
import numpy as np
filepath='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df=pd.read_csv(filepath)
df

```

```python
#shape
df.shape
```

```python
#head
df.head()
```

```python
#column names
df.columns
```

```python
#data types
df.dtypes
```

```python
#missing values
df.isnull().sum()
```

```python
#check duplicates
df.duplicated().sum()
```

```python
#remove duplicates
df=df.drop_duplicates()
df.duplicated().sum()
```

```python
#rename columns
df.rename(columns={"Title":"Movie_Title"},inplace=True)
df.columns
```

```python
#unique genres
df["Genre"].unique()
```

```python
#number of genres
def count_genres(genres):
    return len(genres.split(","))
df["Genre_Count"] = df["Genre"].apply(count_genres)
df[["Movie_Title", "Genre_Count"]]
```

```python
#handle missing values
df["Revenue (Millions)"]=df["Revenue (Millions)"].replace(
    np.nan, df["Revenue (Millions)"].mean()
)

df["Metascore"]=df["Metascore"].replace(
    np.nan, df["Metascore"].mean()
)
```

```python
#top 5 movies
df.sort_values("Rating", ascending=False)[["Movie_Title", "Rating"]].head(5)
```

```python
#movies with highest revenue
df.loc[df["Revenue (Millions)"].idxmax(),["Movie_Title", "Revenue (Millions)"]]
```

```python
#movies with lowest revenue
df.loc[df["Revenue (Millions)"].idxmin(),["Movie_Title", "Revenue (Millions)"]]
```

```python
#avg rating by director
df.groupby("Director")["Rating"].mean()
```

```python
#movies per year
df.groupby("Year")["Movie_Title"].count()
```

```python
#movies with rating > 8 and revenue > 100
df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)][["Movie_Title","Rating","Revenue (Millions)"]]
```

```python
#avg rating and revenue by director
df.groupby("Director")[["Rating","Revenue (Millions)"]].mean()
```

```python
#number of movies and avg rating by year
df.groupby("Year").agg({
    "Movie_Title":"count",
    "Rating":"mean"
})
```

```python
#profitability
df["Profitability"] = (df["Revenue (Millions)"] / df["Runtime (Minutes)"])
```

```python
#top 10 profitable movies
df.sort_values("Profitability",ascending=False)[["Movie_Title","Profitability"]].head(10)
```

```python
#rating category
def rating_category(rating):
   if rating < 6:
    return "Poor"

   elif rating < 7.5:
    return "Average"

   else:
    return "Excellent"

df["Rating_Category"] = df["Rating"].apply(rating_category)
df[["Movie_Title","Rating","Rating_Category"]]
```
